# Avaliação do dataset

```python
import pandas as pd 
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("dataset_1sec_teste.xlsx").drop(columns='Unnamed: 0')
df.iloc[:26]
```

```python
analise_se_cnep = df.loc[[1,26]].drop(columns=['cod_exame','id_paciente'])
analise_se_cnep['diagnostico'].replace({'PNES':0,'SE':1},inplace=True)
analise_se_cnep.reset_index(inplace=True,drop=True)
analise_se_cnep_t = analise_se_cnep.transpose()
analise_se_cnep_t.columns=['PNES','CNEP']
analise_se_cnep_t.drop('diagnostico',inplace=True)
```

```python
analise_se_cnep_t.plot(kind='bar')
```

```python
parametros = df.drop(columns=['id_paciente']).columns
sns.pairplot(data=df[parametros],hue='diagnostico')
```
